```python
# ============================================================
# Sanity Test — small DataFrame + plot
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Test 1: DataFrame creation
test_df = pd.DataFrame({
    "name": ["Alice", "Bob", "Carol"],
    "spend": [100, 250, 175]
})
print("DataFrame test:")
print(test_df)
print()

# Test 2: Groupby (very common operation)
print("Groupby test:")
print(test_df.groupby("name")["spend"].sum())
print()

# Test 3: Plot
plt.figure(figsize=(6, 3))
plt.bar(test_df["name"], test_df["spend"], color="steelblue")
plt.title("Sanity Test Plot")
plt.xlabel("Name")
plt.ylabel("Spend ($)")
plt.show()

print("✅ Everything works. Ready to continue.")
```
